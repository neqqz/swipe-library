```python
# Converts a layout json file to a SpecialDecoder object for layout matching in the Android app
#
# Essentially, all coordinates are normalized [0-1] in X/Y independently, and the offset/scale is also stored.
# In the Android app, we also normalize the keyboard's layout the same way and, if it matches, calculate the
# necessary offset and scale for swipe coordinates to match the layout json.
#
# In standard cases the final scale/offset is unit, but if the layout differs slightly (e.g. number row is active),
# then there will be scaling and offsetting so that the decoder model gets the input it expects.
```

```python
import json
layout = "models/layouts/en_qwerty.json"
layout = json.loads(open(layout, "r").read())
```

```python
min_x = min(key["cx"] for key in layout["keys"])
min_y = min(key["cy"] for key in layout["keys"])
max_x = max(key["cx"] for key in layout["keys"])
max_y = max(key["cy"] for key in layout["keys"])

x_offset = min_x
y_offset = min_y
x_scale = max_x - x_offset
y_scale = max_y - y_offset

key_by_letter = dict((key["letter"], key) for key in layout["keys"])

xs = []
ys = []
letters = layout["letters"]

for letter in letters:
    key = key_by_letter[letter]
    xs.append((key["cx"] - x_offset) / x_scale)
    ys.append((key["cy"] - y_offset) / y_scale)

print(f"""SpecialDecoder(
    layoutLetters = "{letters}",
    layoutXs = listOf({", ".join(str(v)+"f" for v in xs)}),
    layoutYs = listOf({", ".join(str(v)+"f" for v in ys)}),
    xOffset = {x_offset}f,
    xScale = {x_scale}f,
    yOffset = {y_offset}f,
    yScale = {y_scale}f,
)""")
```
